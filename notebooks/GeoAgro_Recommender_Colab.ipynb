```python
from geoagro.data import generate_association_products, load_data
from geoagro.ranking import rank_associations

data = load_data()
offers = generate_association_products(
    tuple(item.id for item in data.associations),
    tuple(item.id for item in data.products),
    seed=42,
)
results = rank_associations(
    data, latitude=-15.7939, longitude=-47.8828,
    max_distance_km=30, season='verão',
    products=['Alface', 'Tomate'], offers=offers,
)
[item.as_dict() for item in results]
```
